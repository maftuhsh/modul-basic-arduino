Struktur Kontrol Percabangan pada Arduino
(If, Else If, Else, Switch)

# Apa itu Percabangan?
Dalam pemrograman Arduino, percabangan adalah cara program mengambil keputusan.

Artinya:
Arduino tidak selalu menjalankan semua perintah secara berurutan,
tapi bisa memilih perintah mana yang dijalankan berdasarkan kondisi tertentu.

Contoh dalam kehidupan sehari-hari:
    - Jika hujan → pakai payung
    - Jika lampu merah → berhenti
    - Jika tidak, lanjut jalan

Logika seperti ini disebut percabangan.

Kenapa Percabangan Penting di Arduino?

Arduino sering digunakan untuk:

- Membaca sensor
- Mengecek tombol
- Mengontrol LED, motor, relay
- Membuat sistem otomatis

Semua itu butuh keputusan, misalnya:
- Apakah suhu terlalu tinggi?
- Apakah tombol ditekan?
- Mode apa yang aktif?

Tanpa percabangan, Arduino tidak bisa “berpikir”.

Konsep Dasar Kondisi (Hal yang biasa Dicek)

Percabangan selalu bekerja dengan kondisi.
Kondisi adalah pernyataan yang hasilnya hanya dua:

- true (benar)
- false (salah)

Contoh kondisi:

suhu > 25
tombol == HIGH
nilai <= 60


If Statement
Apa itu if?

if digunakan untuk mengecek satu kondisi.
    Jika kondisi benar, maka kode di dalam if dijalankan.
    Jika salah, kode dilewati.



```python
// contoh code
int suhu = 30;

void setup() {
  Serial.begin(9600);
  if (suhu > 25) {
    Serial.println("Suhu tinggi");
  }
}

void loop() {
}


// Penjelasan 
// Arduino mengecek: apakah suhu lebih dari 25?
// Karena benar → tampil "Suhu tinggi"
// Jika tidak benar → tidak menampilkan apa-apa
```

Else Statement
Apa itu else?

else adalah jalan alternatif jika kondisi if salah.
Struktur berpikir:

Jika A tidak terjadi, maka lakukan B

```python
// contoh program

int suhu = 20;

void setup() {
  Serial.begin(9600);
  if (suhu > 25) {
    Serial.println("Suhu tinggi");
  } else {
    Serial.println("Suhu normal");
  }
}

void loop() {
}

Penjelasan :
- Jika suhu > 25 → “Suhu tinggi”
- Jika tidak → “Suhu normal”
- Salah satu pasti dijalankan
```

Else If Statement
Apa itu else if?
- Digunakan untuk mengecek banyak kondisi secara berurutan.

Arduino akan:
- Mengecek kondisi pertama
- Jika salah → cek kondisi berikutnya
- Berhenti di kondisi pertama yang benar

```python
int suhu = 15;

void setup() {
  Serial.begin(9600);
  if (suhu > 30) {
    Serial.println("Suhu sangat tinggi");
  } else if (suhu > 25) {
    Serial.println("Suhu tinggi");
  } else if (suhu > 20) {
    Serial.println("Suhu normal");
  } else {
    Serial.println("Suhu rendah");
  }
}

void loop() {
}

// Penjelasan:

// Arduino mengecek dari atas ke bawah
// Saat satu kondisi benar → langsung berhenti
// Urutan kondisi sangat penting
```

Switch Statement
Apa itu switch?

switch digunakan jika:
- Hanya satu variabel
- Nilainya berupa pilihan tetap (1, 2, 3, dll)
- Lebih rapi dibanding if-else panjang.

```python
// contoh program
int hari = 3;

void setup() {
  Serial.begin(9600);
  switch (hari) {
    case 1:
      Serial.println("Senin");
      break;
    case 2:
      Serial.println("Selasa");
      break;
    case 3:
      Serial.println("Rabu");
      break;
    default:
      Serial.println("Hari tidak valid");
  }
}

void loop() {
}

Penjelasan: 

// case → nilai yang dicek
// break → menghentikan pengecekan
// default → jika tidak ada yang cocok

// Jika break tidak ditulis, program akan lanjut ke case berikutnya.
```
