# Apa itu Variabel?

Secara sederhana, variabel adalah lokasi penyimpanan data di memori Arduino yang memiliki nama dan tipe tertentu. Secara Praktis Variabel digunakan untuk menyimpan nilai yang dapat berubah selama program berjalan.

Dalam konteks Arduino, variabel biasa digunakan untuk:
- Menyimpan hasil pembacaan sensor
- Menyimpan status (ON/OFF, ditekan/tidak)
- Menyimpan nilai perhitungan
- Menyimpan teks atau pesan

Komponen Utama Variabel

Setiap variabel memiliki tiga komponen penting:

1. Nama variabel
- Digunakan untuk mengakses data
Contoh: sensorValue

2. Tipe data
- Menentukan jenis dan ukuran data yang disimpan
Contoh: int, float, bool

3. Nilai
- Isi dari variabel tersebut
Contoh: 1023, 3.3, true

contoh deklarasi:
int sensorValue = 500;

Tipe Data Dasar Arduino 

- void
    - Digunakan pada fungsi yang tidak menghasilkan nilai keluaran
    - Umumnya dipakai pada setup() dan loop()

    void setup() {
}

- bool / boolean
    - Menyimpan nilai logika
    - ya memiliki dua kondisi: true atau false
    - Sangat umum dipakai pada percabangan dan kontrol

Contoh:

    bool tombolDitekan = false;

- byte
    - Menyimpan angka bulat kecil (0–255)
    - Efisien dalam penggunaan memori
    - Cocok untuk nilai PWM atau data sederhana

    byte nilaiPWM = 180;


- char
    - Menyimpan satu karakter
    - Menggunakan standar ASCII

    char kode = 'A';

- int
    - Menyimpan angka bulat
    - Paling sering digunakan dalam pemrograman Arduino
    - Pada Arduino UNO: 2 byte

    int jarak = 120;

- long
    - Digunakan untuk angka yang sangat besar
    - Umumnya dipakai untuk waktu atau counter
    - long waktuMillis = 1000000;

- float
    - Menyimpan angka desimal
    - Digunakan pada data sensor analog (suhu, tegangan, dll)

    float suhu = 26.75;

- String
    - Menyimpan data teks
    - Mudah digunakan, namun relatif lebih boros memori
    String pesan = "Arduino Aktif";

```python
// contoh kasus

int sensorValue = 0;
float tegangan = 5.0;
bool sistemAktif = true;
String status = "Siap";

// Interpretasi:
// - sensorValue → nilai sensor
// - tegangan → hasil perhitungan analog
// - sistemAktif → status logika
// - status → informasi teks
```

Ringkasan Tipe Data

| Tipe   | Fungsi Utama       |
| ------ | ------------------ |
| bool   | Logika benar/salah |
| byte   | Angka kecil        |
| int    | Angka bulat        |
| long   | Angka besar        |
| float  | Angka desimal      |
| String | Teks               |
