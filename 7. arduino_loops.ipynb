👉 Itulah perulangan.

Dalam Arduino :
Perulangan adalah cara menyuruh Arduino menjalankan kode yang sama berulang kali tanpa harus menulis ulang kodenya.

Tanpa perulangan → program panjang, boros, dan tidak efisien
Dengan perulangan → program rapi, singkat, dan pintar

Mengapa Perulangan Penting di Arduino?

Arduino bekerja di dunia nyata, misalnya:

- Membaca sensor berulang-ulang
- Mengecek tombol terus-menerus
- Menyalakan LED berkali-kali
- Mengirim data setiap beberapa detik

Tanpa perulangan → Arduino “berhenti bekerja”
Dengan perulangan → Arduino terus hidup dan responsif

Jenis Perulangan di Arduino

Arduino punya 3 jenis perulangan utama:
for
while
do-while

Perbedaannya bukan di hasil, tapi di cara berpikir kapan perulangan berhenti.

Perulangan for
Kapan Digunakan?
Gunakan for jika:
Kamu sudah tahu berapa kali perulangan dilakukan

Cara Berpikir
“Lakukan ini dari angka A sampai angka B.”

bentuk awal:
for (awal; syarat; perubahan) {
  // kode berulang
}


contoh:
for (int i = 0; i < 5; i++) {
  Serial.println(i);
}

penjelasan kode:

Mulai dari i = 0
Selama i < 5, jalankan kode
Setiap selesai satu putaran, i naik 1

ouputnya:
0
1
2
3
4


Perulangan while
Kapan Digunakan?

Gunakan while jika:
Kamu tidak tahu pasti berapa kali perulangan terjadi

Tapi tahu syarat berhentinya

Cara Berpikir
“Selama kondisi ini masih benar, terus lakukan.”

contoh code:

int i = 0;

while (i < 5) {
  Serial.println(i);
  i++;
}

Penjelasan :

Cek dulu: i < 5
Kalau benar → jalankan kode
Jangan lupa menaikkan i, kalau tidak → loop tak berujung

output:
0
1
2
3
4

Contoh Penggunaan Nyata
- Selama tombol belum ditekan
- Selama sensor belum mencapai batas
- Selama koneksi belum terhubung

Perulangan do-while
Kapan Digunakan?

Gunakan do-while jika:
Kamu ingin kode pasti dijalankan minimal satu kali

Cara Berpikir
“Jalankan dulu, nanti baru dicek boleh lanjut atau tidak.”

contoh code:

int i = 0;

do {
  Serial.println(i);
  i++;
} while (i < 5);


Penjelasan Bahasa Awam

- Arduino langsung menjalankan kode
- Setelah itu baru dicek kondisinya
- Jika masih benar → ulangi

Perbedaan Penting

Jika kondisi salah sejak awal:

while → tidak jalan sama sekali

do-while → tetap jalan 1 kali

Kesimpulan:
| Perulangan | Dicek Kapan   | Cocok Untuk      |
| ---------- | ------------- | ---------------- |
| `for`      | Sebelum jalan | Jumlah pasti     |
| `while`    | Sebelum jalan | Kondisi tertentu |
| `do-while` | Setelah jalan | Minimal 1 kali   |
